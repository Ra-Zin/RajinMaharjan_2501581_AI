```python
def r2(y, y_pred):

    mean_y = np.mean(y)
    ss_tot = np.sum((y - mean_y) **  2)
    ss_res = np.sum((y - y_pred)  ** 2)
    r2_score = 1 - (ss_res / ss_tot)
    return r2_score
```
