```python
# Inspect dataset for numeric column stored as text
for col in hdi_2022_df.columns:
    if hdi_2022_df[col].dtype == "object":
        print(f"\nChecking column: {col}")
        print(f"Sample values: {hdi_2022_df[col].head().tolist()}")
        
        has_letters = False
        
        for value in hdi_2022_df[col].dropna().astype(str):
            for char in value.replace('.', '').replace('-', '').replace('+', '').strip():
                if char.isalpha():  # If we find a letter
                    print(f"  Found letter '{char}' in value: {value}")
                    has_letters = True
                    break
            if has_letters:
                break
        
        if not has_letters:
            print(f"{col} is numeric but stored as text")
        else:
            print(f"{col} contains letter, so skipping")
print("\n")
print("All the other numeric columns are already stored as int or float")
```

```python
# Use the 2022 subset dataframe
country_names = hdi_2022_df["country"].unique()
print(f"Total unique countries: {len(country_names)}")

# Case-insensitive version for checking duplicates
from collections import Counter

country_lower = [str(name).lower().strip() for name in country_names]
duplicate_check = Counter(country_lower)

potential_duplicates = {k: v for k, v in duplicate_check.items() if v > 1}

if potential_duplicates:
    print("Potential duplicate country names (case-insensitive):")
    for country, count in potential_duplicates.items():
        matching = [name for name in country_names
                    if str(name).lower().strip() == country]
        print(f"  '{country}': {matching}")
else:
    print("No obvious duplicate country names found")
```

```python
# Inspect dublicate rows
print("Checking for duplicate rows:")
# Exact duplicates
exact_duplicates = hdi_2022_df.duplicated().sum()
print(f"  Exact duplicate rows: {exact_duplicates}")
```

```python
# Inspect for special characters representing missing data
print("Checking for special characters representing missing data:")
special_chars = ['--', '–', '---', 'na', 'NA', 'N/A', 'n/a', 'null', 'NULL', '?', '??', '???']
special_char_found = False

for col in hdi_2022_df.select_dtypes(include=['object']).columns:
    if hdi_2022_df[col].notna().any():  # Check only if column has non-null values
        for char in special_chars:
            # Using regex=False to treat the string as a literal, not a regex pattern
            count = hdi_2022_df[col].astype(str).str.contains(char, na=False, regex=False).sum()
            if count > 0:
                print(f"'{col}': Found {count} instances of '{char}'")
                special_char_found = True

if not special_char_found:
    print("No special characters representing missing data found")
```

```python
print("HANDLING DUPLICATES")

# Remove exact duplicate rows
initial_rows = len(hdi_2022_df)
hdi_2022_df = hdi_2022_df.drop_duplicates()
removed_exact = initial_rows - len(hdi_2022_df)
print(f"Removed {removed_exact} exact duplicate rows\n")

# Handling missing values
missing_before = hdi_2022_df.isnull().sum()
print("Missing values before handling:")
print(missing_before[missing_before > 0])

for col in hdi_2022_df.columns:
    missing_count = hdi_2022_df[col].isnull().sum()
    if missing_count > 0:
        if hdi_2022_df[col].dtype in ["float64", "int64"]:
            # Numeric: impute with median
            impute_value = hdi_2022_df[col].median()
            hdi_2022_df[col] = hdi_2022_df[col].fillna(impute_value)
            print(f"  - {col}: Imputed {missing_count} missing values with median ({impute_value:.2f})")
        elif hdi_2022_df[col].dtype == "object":
            # Categorical: impute with mode
            if hdi_2022_df[col].notna().sum() > 0:
                mode_vals = hdi_2022_df[col].mode()
                impute_value = mode_vals[0] if not mode_vals.empty else "Unknown"
                hdi_2022_df[col] = hdi_2022_df[col].fillna(impute_value)
                print(f"  - {col}: Imputed {missing_count} missing values with mode ('{impute_value}')")
```

3. Outlier Detection

```python
# Detect outliers in HDI and GNI per Capita using the 1.5 × IQR rule.

# Function to detect outliers using 1.5 * IQR rule
def detect_outliers(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return (data[column] < lower_bound) | (data[column] > upper_bound)

# Detect outliers
South_Asia_latest['hdi_outlier'] = detect_outliers(South_Asia_latest, 'hdi')
South_Asia_latest['gni_outlier'] = detect_outliers(South_Asia_latest, 'gross_inc_percap')
South_Asia_latest['is_outlier'] = South_Asia_latest['hdi_outlier'] | South_Asia_latest['gni_outlier']
```
